```python
import pandas as pd
import numpy as np
import torch
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

lenta = pd.read_csv("https://github.com/mannefedov/compling_nlp_hse_course/raw/master/data/lenta_40k.csv.zip").sample(500, random_state=8)

model = SentenceTransformer('cointegrated/rubert-tiny2')

texts = lenta['text'].fillna('').astype(str).tolist()

embeddings = model.encode(
    texts,
    convert_to_numpy=True,
    show_progress_bar=True
)

similarities = cosine_similarity(embeddings)

np.fill_diagonal(similarities, -np.inf)

i, j = np.unravel_index(np.argmax(similarities), similarities.shape)

first_word_1 = texts[i].split()[0]
first_word_2 = texts[j].split()[0]

print(first_word_1, first_word_2)
```
